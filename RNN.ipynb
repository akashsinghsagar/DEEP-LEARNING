```python
sentence = ["I", "love", "deep", "learning"]

word_value = {
    "I": 1,
    "love": 2,
    "deep": 3,
    "learning": 4
}

# convert words to numbers
sequence = [word_value[word] for word in sentence]

print("Input sequence:", sentence)
print("Encode sequence:", sequence)

# STEP 2: Define the RNN model
w_input = 0.5
w_memory = 0.8
bias = 0.0

# STEP 3: Initialize the memory state
memory = 0.0

# STEP 4: Process the input sequence through the RNN
print("\nProcessing sequence through RNN:")

for t in range(len(sequence)):
    input_value = sequence[t]
    # Compute the new memory state before using it in prints
    new_memory = w_input * input_value + w_memory * memory + bias
    
    
    print(f"Time step {t+1}:")
    print(f"  Input value: {input_value}")
    print(f"  Previous memory: {memory:.2f}")
    print(f"  New memory: {new_memory:.2f}")

    print("New Memory Calculation:")
    print(f"  w_input * input_value: {w_input} * {input_value} = {w_input * input_value:.2f}")

    memory = new_memory

    print(f"  Updated memory: {memory:.2f}\n")

# STEP 5: Final output
print("Final memory state after processing the sequence:", memory)

# decision
if memory > 5:
    print("The model predicts a positive outcome.")
else:
    print("Bad")
```
