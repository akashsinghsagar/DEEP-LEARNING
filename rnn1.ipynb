```python
# Improved Simple RNN Example
sentence = ["I", "love", "deep", "learning"]
word_value = {
    "I": 1,
    "love": 2,
    "deep": 3,
    "learning": 4
}

# Convert words to numbers
sequence = [word_value[word] for word in sentence]

print("Input sequence:", sentence)
print("Encoded sequence:", sequence)

# STEP 2: Define the RNN model parameters
w_input = 0.5
w_memory = 0.8
bias = 0.0

# STEP 3: Initialize the memory state
memory = 0.0

# STEP 4: Process the input sequence through the RNN
print("\nProcessing sequence through RNN:")

for t, input_value in enumerate(sequence):
    # Compute the new memory state
    new_memory = w_input * input_value + w_memory * memory + bias
    
    print(f"Time step {t+1}:")
    print(f"  Input value: {input_value}")
    print(f"  Previous memory: {memory:.2f}")
    print(f"  New memory: {new_memory:.2f}")
    print("  Calculation:")
    print(f"    w_input * input_value: {w_input} * {input_value} = {w_input * input_value:.2f}")
    print(f"    w_memory * memory: {w_memory} * {memory:.2f} = {w_memory * memory:.2f}")
    print(f"    bias: {bias}")
    print(f"    Total: {new_memory:.2f}")

    # Update memory for next step
    memory = new_memory
    print(f"  Updated memory: {memory:.2f}\n")

# STEP 5: Final output
print("Final memory state after processing the sequence:", memory)

# Decision based on final memory state
if memory > 5:
    print("The model predicts a positive outcome.")
else:
    print("The model predicts a negative outcome.")
```

# RNN Visualization and Analysis
This section visualizes the memory state evolution and provides further analysis of the simple RNN.

```python
# Visualize memory state evolution
import matplotlib.pyplot as plt

# Re-run the RNN and store memory states
sentence = ["I", "love", "deep", "learning"]
word_value = {"I": 1, "love": 2, "deep": 3, "learning": 4}
sequence = [word_value[word] for word in sentence]
w_input = 0.5
w_memory = 0.8
bias = 0.0
memory = 0.0
memory_states = [memory]

for input_value in sequence:
    new_memory = w_input * input_value + w_memory * memory + bias
    memory = new_memory
    memory_states.append(memory)

plt.figure(figsize=(8,4))
plt.plot(range(len(memory_states)), memory_states, marker='o')
plt.title('RNN Memory State Evolution')
plt.xlabel('Time Step')
plt.ylabel('Memory State')
plt.grid(True)
plt.xticks(range(len(memory_states)), ['Start'] + sentence)
plt.show()
```
